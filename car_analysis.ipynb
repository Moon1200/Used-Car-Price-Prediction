```python
import pandas as pd
df = pd.read_csv("clean_cars.csv")
print(df.shape)  
```

```python
from scipy import stats

auto = df[df['Gear_Type']=='Automatic']['Price']
manual = df[df['Gear_Type']=='Manual']['Price']
t_stat, p_val = stats.ttest_ind(auto, manual, equal_var=False)
print(f"متوسط أوتوماتيك: {auto.mean():.0f}, متوسط عادي: {manual.mean():.0f}")
print(f"T={t_stat:.3f}, p-value={p_val:.6f}")
```

```python
groups = [g['Price'].values for name, g in df.groupby('Origin')]
f_stat, p_val2 = stats.f_oneway(*groups)
print(f"F={f_stat:.3f}, p-value={p_val2:.6f}")
```

```python
top_makes = df['Make'].value_counts().head(8).index
df_model = df[df['Make'].isin(top_makes)].copy()

import statsmodels.formula.api as smf

model = smf.ols(
    "Price ~ Year + Mileage + Engine_Size + C(Make) + C(Gear_Type) + C(Origin)",
    data=df_model
).fit()

print(model.summary())
print(f"\nR-squared: {model.rsquared:.3f}")
```

```python
model.params.to_csv("car_price_coefficients.csv")
```
